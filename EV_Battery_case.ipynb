### Battery Sensor Data Analysis

```python
import pandas as pd

# Battery sensor data from the description
data = {
    'Time (h)': [0, 1, 2, 3, 4, 5],
    'Temperature (°C)': [32, 35, 38, 40, 42, 45],
    'Voltage (V)': [400, 395, 390, 385, 380, 375],
    'Current (A)': [20, 22, 25, 28, 30, 32],
    'Charge (%)': [95, 88, 80, 70, 60, 48]
}

df_battery = pd.DataFrame(data)

# Display the DataFrame
display(df_battery)
```

#Task 1: Analysis Using Python Basics

#### 1. Display the sensor readings using a for loop

```python
print("Battery Sensor Readings:")
for index, row in df_battery.iterrows():
    print(f"Time: {row['Time (h)']}h, Temp: {row['Temperature (°C)']}°C, Voltage: {row['Voltage (V)']}V, Current: {row['Current (A)']}A, Charge: {row['Charge (%)']}% ")
```

#### 2. Identify high temperature

```python
print("Temperature Status:")
for index, row in df_battery.iterrows():
    if row['Temperature (°C)'] > 40:
        print(f"Time {row['Time (h)']}h: High Temperature Alert ({row['Temperature (°C)']}°C)")
    else:
        print(f"Time {row['Time (h)']}h: Temperature Normal ({row['Temperature (°C)']}°C)")
```

#### 3. Identify low battery

```python
print("Battery Charge Status:")
for index, row in df_battery.iterrows():
    if row['Charge (%)'] < 50:
        print(f"Time {row['Time (h)']}h: Low Battery Alert ({row['Charge (%)']}%) ")
    else:
        print(f"Time {row['Time (h)']}h: Battery Charge Normal ({row['Charge (%)']}%) ")
```

#### 4. Create a function `battery_status(charge)`

```python
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"

print("Battery Status using function:")
for index, row in df_battery.iterrows():
    status = battery_status(row['Charge (%)'])
    print(f"Time {row['Time (h)']}h: Charge {row['Charge (%)']}% - Status: {status}")
```

### Task 3: Analysis Using Linear Algebra

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
X = np.array([[40], [25]])

Y = np.dot(A, X)
print("Resulting vector Y (Battery Stress Index):")
display(Y)
```

#### 1. Transpose of A (Aᵀ)

```python
A_transpose = A.T
print("Transpose of A (Aᵀ):")
display(A_transpose)
```

#### 2. Determinant of A (det(A))

```python
A_determinant = np.linalg.det(A)
print(f"Determinant of A (det(A)): {A_determinant}")
```

#### 3. Inverse of A (A⁻¹)

```python
try:
    A_inverse = np.linalg.inv(A)
    print("Inverse of A (A⁻¹):")
    display(A_inverse)
except np.linalg.LinAlgError:
    print("Inverse does not exist for matrix A.")
```

#### Solving a System of Equations

```python
A_system = np.array([[2, 1], [1, 3]])
B_system = np.array([100, 120])

solution = np.linalg.solve(A_system, B_system)

print(f"Solution for x: {solution[0]:.2f}")
print(f"Solution for y: {solution[1]:.2f}")
```

### Task 4: Analysis Using Calculus

#### 1. Find the rate of change of battery charge

```python
def C_t(t):
    return 100 - 2 * t - 0.5 * t**2

def dC_dt(t):
    return -2 - t

print("Battery charge function: C(t) = 100 - 2t - ½t²")
print("Rate of change of battery charge: dC/dt = -2 - t")
```

#### 2. Calculate the discharge rate

```python
discharge_rate_t2 = dC_dt(2)
discharge_rate_t5 = dC_dt(5)

print(f"Discharge rate at t = 2 hours (dC/dt): {discharge_rate_t2} %/h")
print(f"Discharge rate at t = 5 hours (dC/dt): {discharge_rate_t5} %/h")
```

#### 3. Interpret the result

```python
if discharge_rate_t2 < discharge_rate_t5:
    interpretation = "The battery is losing charge at an increasing rate."
elif discharge_rate_t2 > discharge_rate_t5:
    interpretation = "The battery is losing charge at a decreasing rate."
else:
    interpretation = "The battery is losing charge at a constant rate."

print(interpretation)
print("Since the derivative dC/dt = -2 - t, as 't' increases, dC/dt becomes more negative, indicating an increasing rate of discharge.")
```

### Generated Dataset for Calculus Analysis

```python
import pandas as pd
import numpy as np

time_points = np.arange(0, 6, 1)

charge_values = [C_t(t) for t in time_points]
discharge_rates = [dC_dt(t) for t in time_points]

calculus_df = pd.DataFrame({
    'Time (h)': time_points,
    'Charge C(t) (%)': charge_values,
    'dC/dt (%/h)': discharge_rates
})

display(calculus_df)
```
