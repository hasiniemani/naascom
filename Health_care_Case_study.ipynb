## 1. Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Please provide the data. Once the data is provided (e.g., as a CSV file), I can load it into a DataFrame. For example, if the data is in `patient_data.csv`:

```python
# Patient monitoring data provided by the user
data = {
    'Time (h)': [0, 1, 2, 3, 4, 5],
    'Heart Rate (BPM)': [76, 82, 90, 98, 105, 112],
    'Body Temperature (°C)': [36.7, 37.0, 37.4, 38.0, 38.5, 39.0],
    'Blood Oxygen Level (SpO₂ %)': [98, 97, 96, 94, 92, 89],
    'Systolic Blood Pressure (mmHg)': [122, 125, 130, 135, 140, 148],
    'Patient Health Score (%)': [92, 88, 82, 74, 65, 52]
}

df = pd.DataFrame(data)
display(df.head())
```

## Task 1 — Analysis Using Python Basics

### 1. Display the Patient Readings
Using a for loop to display all patient readings.

```python
print("Patient Readings:")
for index, row in df.iterrows():
    print(f"Time: {row['Time (h)']}h, Heart Rate: {row['Heart Rate (BPM)']} BPM, Temperature: {row['Body Temperature (°C)']}°C, SpO₂: {row['Blood Oxygen Level (SpO₂ %)']}%, Systolic BP: {row['Systolic Blood Pressure (mmHg)']} mmHg, Health Score: {row['Patient Health Score (%)']}%")
```

### 2. Identify High Temperature
If Temperature > 38°C, display “High Temperature Alert”. Otherwise, display “Temperature Normal”.

```python
print("\nTemperature Status:")
for index, row in df.iterrows():
    if row['Body Temperature (°C)'] > 38:
        print(f"Time {row['Time (h)']}h: High Temperature Alert ({row['Body Temperature (°C)']}°C)")
    else:
        print(f"Time {row['Time (h)']}h: Temperature Normal ({row['Body Temperature (°C)']}°C)")
```

### 3. Identify Low Oxygen Level
If SpO₂ < 90%, display “Low Oxygen Alert”. Otherwise, display “Oxygen Level Normal”.

```python
print("\nOxygen Level Status:")
for index, row in df.iterrows():
    if row['Blood Oxygen Level (SpO₂ %)'] < 90:
        print(f"Time {row['Time (h)']}h: Low Oxygen Alert ({row['Blood Oxygen Level (SpO₂ %)']}%) ")
    else:
        print(f"Time {row['Time (h)']}h: Oxygen Level Normal ({row['Blood Oxygen Level (SpO₂ %)']}%) ")
```

### 4. Create a Function
Create `health_status(score)` that returns: Healthy (score > 80), Monitor (50 ≤ score ≤ 80), Critical (score < 50).

```python
def health_status(score):
    if score > 80:
        return "Healthy"
    elif 50 <= score <= 80:
        return "Monitor"
    else:
        return "Critical"

print("\nPatient Health Status:")
for index, row in df.iterrows():
    status = health_status(row['Patient Health Score (%)'])
    print(f"Time {row['Time (h)']}h: Health Score {row['Patient Health Score (%)']}% -> Status: {status}")
```

## Task 2 — Analysis Using NumPy
