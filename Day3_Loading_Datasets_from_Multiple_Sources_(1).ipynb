The `oncologists-in-new-york.zip` file has been extracted to the `dataset` directory. Let's list the contents of this directory to find the actual data file to load.

````python
import os
import zipfile

# Re-download the zip file to ensure it exists
!kaggle datasets download -d tamashasfernando/oncologists-in-new-york --force

# Extract the zip file to ensure the 'dataset' directory exists
with zipfile.ZipFile("oncologists-in-new-york.zip", 'r') as zip_ref:
    zip_ref.extractall("dataset")

# List the contents of the 'dataset' directory
print(os.listdir('dataset'))
````

Now that we know the CSV file is `oncologists_nyc.csv` inside the `dataset` directory, let's load it into a pandas DataFrame.

````python
import pandas as pd
import os

# Construct the full path to the CSV file
csv_file_path = os.path.join('dataset', 'oncologists_nyc.csv')

# Load the CSV file into a pandas DataFrame
df_oncologists = pd.read_csv(csv_file_path)

# Display the first few rows of the DataFrame to confirm loading
print(df_oncologists.head())
````

First, let's look at the basic information about the DataFrame, such as column names, non-null counts, and data types, using `df.info()`.

````python
# Get a concise summary of the DataFrame
df_oncologists.info()
````

Next, let's see the dimensions of the dataset (number of rows and columns) using the `df.shape` attribute.

````python
# Display the shape of the DataFrame
print(f"Dataset shape: {df_oncologists.shape}")
````

For numerical columns, `df.describe()` provides a statistical summary including count, mean, standard deviation, min, max, and quartile values. This helps in understanding the distribution of numerical data.

````python
# Generate descriptive statistics for numerical columns
df_oncologists.describe()
````

It's also important to check for missing values in each column to understand data completeness. We can do this with `df.isnull().sum()`.

````python
# Check for missing values in each column
print("Missing values per column:\n", df_oncologists.isnull().sum())
````

````python
# Display the first five rows of the DataFrame
display(df_oncologists.head())
````

### Extracting Data for a Specific Topic: 'Specialty'

Let's extract data related to the 'Specialty' column to understand the different types of oncologists present in the dataset.

````python
# Get all unique specialties (using 'categoryName' column)
unique_specialties = df_oncologists['categoryName'].unique()
print("Unique Specialties:", unique_specialties)
````

````python
# Get the count of each specialty (using 'categoryName' column)
specialty_counts = df_oncologists['categoryName'].value_counts()
print("\nSpecialty Counts:\n", specialty_counts)
````

Now, let's say we want to extract all oncologists who specialize in 'Hematology/Oncology'. We can filter the DataFrame based on this condition.

````python
# Filter the DataFrame for oncologists specializing as 'Oncologist' (using 'categoryName' column)
# Note: 'Hematology/Oncology' was not found as a direct category, so 'Oncologist' is used as an example.
oncologists_filtered = df_oncologists[df_oncologists['categoryName'] == 'Oncologist']

print("\nOncologists specializing as 'Oncologist':")
display(oncologists_filtered.head())
````

#Step 5: Loading Dataset from Public API

Example: Exchange Rate API

````python
import requests
import pandas as pd

url = "https://api.exchangerate-api.com/v4/latest/USD"

response = requests.get(url)
data = response.json()

rates = data["rates"]

df = pd.DataFrame(list(rates.items()), columns=["Currency","Rate"])
df.head()
````

#Step 6: Loading Dataset from SQL Database

Create sample database.

````python
import sqlite3
import pandas as pd

conn = sqlite3.connect("bank.db")

data = {
    "Customer_ID":[1,2,3,4],
    "Loan_Amount":[200000,150000,300000,180000],
    "Credit_Score":[720,680,650,700]
}

df = pd.DataFrame(data)

df.to_sql("customers", conn, if_exists="replace", index=False)
````

Query Database

````python
query = "SELECT * FROM customers"

df = pd.read_sql(query, conn)
df.head()
````
