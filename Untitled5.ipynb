```python
"""NumPy Linear Algebra — Examples and Exercises
===============================================

This file walks through common NumPy linear algebra operations.
Each section has:
  1. A short explanation
  2. A working example program
  3. Exercises for you to try (write your solution code right below
     the exercise, in the marked space)

Run this file directly to see all example outputs:
    python numpy_linear_algebra.py
"""

import numpy as np


def section_header(title):
    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)


# ---------------------------------------------------------------
# 1. Creating Matrices and Vectors
# ---------------------------------------------------------------
def example_create_arrays():
    section_header("1. Creating Matrices and Vectors")

    v = np.array([1, 2, 3])
    A = np.array([[1, 2], [3, 4]])
    I = np.eye(3)          # identity matrix
    Z = np.zeros((2, 3))
    O = np.ones((3, 2))

    print("Vector v:\n", v)
    print("Matrix A:\n", A)
    print("Identity I:\n", I)
    print("Zeros Z:\n", Z)
    print("Ones O:\n", O)


# Exercises:
# 1a. Create a 4x4 identity matrix and print it.
# 1b. Create a 3x3 matrix filled with the number 7 (hint: np.full).
# 1c. Create a vector of 10 evenly spaced numbers between 0 and 1
#     (hint: np.linspace).
#
# --- your code here ---
identity_matrix = np.eye(4)
print("4x4 Identity Matrix:\n", identity_matrix)

matrix_of_sevens = np.full((3, 3), 7)
print("\n3x3 Matrix filled with 7:\n", matrix_of_sevens)

vector = np.linspace(0, 1, 10)
print("\n10 evenly spaced numbers between 0 and 1:\n", vector)

# ---------------------------------------------------------------
# 2. Matrix Multiplication
# ---------------------------------------------------------------
def example_matrix_multiplication():
    section_header("2. Matrix Multiplication")

    A = np.array([[1, 2], [3, 4]])
    B = np.array([[5, 6], [7, 8]])

    elementwise = A * B          # element-wise multiplication
    matmul_operator = A @ B      # matrix multiplication (preferred)
    matmul_function = np.matmul(A, B)
    dot_function = np.dot(A, B)

    print("A:\n", A)
    print("B:\n", B)
    print("Element-wise A*B:\n", elementwise)
    print("Matrix mult A@B:\n", matmul_operator)
    print("np.matmul(A, B):\n", matmul_function)
    print("np.dot(A, B):\n", dot_function)


# Exercises:
# 2a. Given A = [[2, 0], [0, 2]] and B = [[1, 2], [3, 4]], compute A@B
#     by hand, then verify with NumPy.
# 2b. Multiply a 3x2 matrix by a 2x4 matrix. What shape is the result?
# 2c. Explain (in a comment) why A*B and A@B give different results.
#
# --- your code here ---

A = np.array([[2, 0],
              [0, 2]])

B = np.array([[1, 2],
              [3, 4]])

result = A @ B

print("A @ B:")
print(result)

C = np.random.randint(1, 10, (3, 2))
D = np.random.randint(1, 10, (2, 4))

result_shape = C @ D

print("\nShape of (3x2) @ (2x4) matrix:")
print(result_shape.shape)


# ---------------------------------------------------------------
# 3. Transpose
# ---------------------------------------------------------------
def example_transpose():
    section_header("3. Transpose")

    A = np.array([[1, 2, 3], [4, 5, 6]])
    print("A:\n", A)
    print("A.T (transpose):\n", A.T)
    print("np.transpose(A):\n", np.transpose(A))


# Exercises:
# 3a. Create a 3x5 matrix and confirm its transpose has shape (5, 3).
# 3b. Verify that (A @ B).T equals B.T @ A.T for two matrices A, B.
#
# --- your code here ---


matrix_3x5 = np.random.randint(1, 10, (3, 5))

transpose_matrix = matrix_3x5.T

print("Original Matrix (3x5):")
print(matrix_3x5)

print("\nTransposed Matrix:")
print(transpose_matrix)

print("\nShape of Transpose:")
print(transpose_matrix.shape)

A = np.random.randint(1, 10, (2, 3))
B = np.random.randint(1, 10, (3, 4))

left_side = (A @ B).T
right_side = B.T @ A.T

print("\n(A @ B).T:")
print(left_side)

print("\nB.T @ A.T:")
print(right_side)

print("\nAre they equal?")
print(np.array_equal(left_side, right_side))

# ---------------------------------------------------------------
# 4. Determinant
# ---------------------------------------------------------------
def example_determinant():
    section_header("4. Determinant")

    A = np.array([[4, 2], [7, 6]])
    det = np.linalg.det(A)
    print("A:\n", A)
    print("Determinant of A:", det)


# Exercises:
# 4a. Compute the determinant of a 3x3 matrix of your choice.
# 4b. Create a singular (non-invertible) matrix and show its
#     determinant is (approximately) zero.
#
# --- your code here ---

matrix_3x3 = np.array([
    [2, 3, 1],
    [4, 1, 5],
    [6, 2, 3]
])

determinant = np.linalg.det(matrix_3x3)

print("3x3 Matrix:")
print(matrix_3x3)

print("\nDeterminant:")
print(determinant)


singular_matrix = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
])

singular_determinant = np.linalg.det(singular_matrix)

print("\nSingular Matrix:")
print(singular_matrix)

print("\nDeterminant of Singular Matrix:")
print(singular_determinant)

print("\nIs determinant approximately zero?")
print(np.isclose(singular_determinant, 0))

# ---------------------------------------------------------------
# 5. Matrix Inverse
# ---------------------------------------------------------------
def example_inverse():
    section_header("5. Matrix Inverse")

    A = np.array([[4, 7], [2, 6]])
    A_inv = np.linalg.inv(A)
    identity_check = A @ A_inv

    print("A:\n", A)
    print("Inverse of A:\n", A_inv)
    print("A @ A_inv (should be identity):\n", identity_check)


# Exercises:
# 5a. Try to invert a singular matrix and observe the error NumPy
#     raises (hint: wrap in try/except).
# 5b. Verify that (A_inv)_inv == A for a matrix of your choice.
#
# --- your code here ---

singular_matrix = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
])

try:
    inverse_matrix = np.linalg.inv(singular_matrix)
    print("Inverse of Singular Matrix:")
    print(inverse_matrix)

except np.linalg.LinAlgError:
    print("Error: Matrix is singular and does not have an inverse.")


A = np.array([
    [4, 7],
    [2, 6]
])

# Find inverse of A
A_inv = np.linalg.inv(A)

# Find inverse of A_inv
A_double_inv = np.linalg.inv(A_inv)

print("\nOriginal Matrix A:")
print(A)

print("\nInverse of A:")
print(A_inv)

print("\nInverse of A_inv:")
print(A_double_inv)

print("\nIs (A_inv)_inv equal to A?")
print(np.allclose(A_double_inv, A))
# ---------------------------------------------------------------
# 6. Solving Systems of Linear Equations
# ---------------------------------------------------------------
def example_solve_linear_system():
    section_header("6. Solving Linear Systems (Ax = b)")

    # 2x + 3y = 8
    # 5x + 4y = 13
    A = np.array([[2, 3], [5, 4]])
    b = np.array([8, 13])

    x = np.linalg.solve(A, b)
    print("A:\n", A)
    print("b:", b)
    print("Solution x:", x)
    print("Check A @ x == b:", np.allclose(A @ x, b))


# Exercises:
# 6a. Solve the system:
#       x + y + z = 6
#       2y + 5z = -4
#       2x + 5y - z = 27
# 6b. Compare np.linalg.solve(A, b) with computing
#     np.linalg.inv(A) @ b. Which is generally preferred, and why?
#
# --- your code here ---

A = np.array([
    [1, 1, 1],
    [0, 2, 5],
    [2, 5, -1]
])

b = np.array([6, -4, 27])

# Solve using np.linalg.solve()
solution = np.linalg.solve(A, b)

print("Solution [x, y, z]:")
print(solution)



solution_using_inverse = np.linalg.inv(A) @ b

print("\nSolution using inverse method:")
print(solution_using_inverse)

print("\nAre both solutions equal?")
print(np.allclose(solution, solution_using_inverse))

# Explanation:
# np.linalg.solve(A, b) is generally preferred because it is:
# - Faster (does not explicitly compute the inverse)
# - More numerically stable
# - More efficient for solving linear systems
#
# Computing np.linalg.inv(A) @ b is less preferred because finding
# the inverse adds extra computation and can introduce more numerical errors.


# ---------------------------------------------------------------
# 7. Eigenvalues and Eigenvectors
# ---------------------------------------------------------------
def example_eigen():
    section_header("7. Eigenvalues and Eigenvectors")

    A = np.array([[4, -2], [1, 1]])
    eigenvalues, eigenvectors = np.linalg.eig(A)

    print("A:\n", A)
    print("Eigenvalues:", eigenvalues)
    print("Eigenvectors (columns):\n", eigenvectors)


# Exercises:
# 7a. Verify A @ v == lambda * v for each eigenvalue/eigenvector pair.
# 7b. Find the eigenvalues of a diagonal matrix and explain the
#     pattern you notice.
#
# --- your code here ---


A = np.array([
    [4, 1],
    [2, 3]
])

# Find eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nVerifying A @ v = lambda * v:")

for i in range(len(eigenvalues)):
    eigenvalue = eigenvalues[i]
    eigenvector = eigenvectors[:, i]

    left_side = A @ eigenvector
    right_side = eigenvalue * eigenvector

    print(f"\nEigenvalue: {eigenvalue}")
    print("A @ v:")
    print(left_side)
    print("lambda * v:")
    print(right_side)
    print("Equal:", np.allclose(left_side, right_side))


diagonal_matrix = np.diag([5, 10, 15])

diagonal_eigenvalues = np.linalg.eigvals(diagonal_matrix)

print("\nDiagonal Matrix:")
print(diagonal_matrix)

print("\nEigenvalues:")
print(diagonal_eigenvalues)

#Explanation--------------

# For a diagonal matrix, the eigenvalues are exactly the diagonal elements.
# The pattern is that each diagonal value directly represents an eigenvalue.`


# ---------------------------------------------------------------
# 8. Norms (vector and matrix length/magnitude)
# ---------------------------------------------------------------
def example_norms():
    section_header("8. Norms")

    v = np.array([3, 4])
    l2_norm = np.linalg.norm(v)          # Euclidean norm
    l1_norm = np.linalg.norm(v, ord=1)   # Manhattan norm

    print("v:", v)
    print("L2 norm (Euclidean):", l2_norm)
    print("L1 norm (Manhattan):", l1_norm)


# Exercises:
# 8a. Compute the L2 norm of a 5-dimensional vector.
# 8b. Normalize a vector (divide by its own norm) so its length is 1,
#     and verify the result with np.linalg.norm.
#
# --- your code here ---

vector_5d = np.array([3, 4, 5, 6, 7])

l2_norm = np.linalg.norm(vector_5d)

print("5-dimensional Vector:")
print(vector_5d)

print("\nL2 Norm:")
print(l2_norm)

normalized_vector = vector_5d / l2_norm

print("\nNormalized Vector:")
print(normalized_vector)

print("\nNorm of Normalized Vector:")
print(np.linalg.norm(normalized_vector))

# The result should be approximately 1
print("\nIs the normalized vector length 1?")
print(np.isclose(np.linalg.norm(normalized_vector), 1))


# ---------------------------------------------------------------
# 9. Rank of a Matrix
# ---------------------------------------------------------------
def example_rank():
    section_header("9. Matrix Rank")

    full_rank = np.array([[1, 2], [3, 4]])
    deficient_rank = np.array([[1, 2], [2, 4]])  # second row = 2 * first row

    print("Full-rank matrix:\n", full_rank, "-> rank =", np.linalg.matrix_rank(full_rank))
    print("Rank-deficient matrix:\n", deficient_rank, "-> rank =", np.linalg.matrix_rank(deficient_rank))


# Exercises:
# 9a. Construct a 4x4 matrix with rank 2.
# 9b. Explain (in a comment) the relationship between rank and
#     whether a matrix is invertible.
#
# --- your code here ---


rank_2_matrix = np.array([
    [1, 2, 3, 4],
    [0, 1, 2, 3],
    [1, 3, 5, 7],   # Row1 + Row2
    [2, 5, 8, 11]   # Row1 + 3*Row2
])

rank = np.linalg.matrix_rank(rank_2_matrix)

print("4x4 Matrix:")
print(rank_2_matrix)

print("\nRank of Matrix:")
print(rank)

#Explanation for 9b--------------

# A square matrix is invertible only if its rank equals its size.
# For an n x n matrix, rank = n means the matrix is full rank and invertible.
# If rank < n, the matrix is singular and does not have an inverse.

# ---------------------------------------------------------------
# 10. Singular Value Decomposition (SVD)
# ---------------------------------------------------------------
def example_svd():
    section_header("10. Singular Value Decomposition (SVD)")

    A = np.array([[1, 2], [3, 4], [5, 6]])
    U, S, Vt = np.linalg.svd(A)

    print("A:\n", A)
    print("U:\n", U)
    print("Singular values S:", S)
    print("V^T:\n", Vt)

    # Reconstruct A from U, S, Vt to verify decomposition
    S_full = np.zeros(A.shape)
    np.fill_diagonal(S_full, S)
    A_reconstructed = U @ S_full @ Vt
    print("Reconstructed A:\n", A_reconstructed)


# Exercises:
# 10a. Reconstruct A from its SVD components and confirm it matches
#      the original (use np.allclose).
# 10b. Use SVD to compute a low-rank approximation of A by keeping
#      only the largest singular value. How close is it to A?
#
# --- your code here ---

A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 10]
])

# Compute SVD
U, S, Vt = np.linalg.svd(A)

# Convert singular values into a diagonal matrix
S_matrix = np.diag(S)

# Reconstruct A
A_reconstructed = U @ S_matrix @ Vt

print("Original Matrix A:")
print(A)

print("\nReconstructed Matrix A:")
print(A_reconstructed)

print("\nDoes reconstructed matrix match A?")
print(np.allclose(A, A_reconstructed))


S_low_rank = np.zeros_like(S_matrix)
S_low_rank[0, 0] = S_matrix[0, 0]

A_low_rank = U @ S_low_rank @ Vt

print("\nRank-1 Approximation of A:")
print(A_low_rank)


error = np.linalg.norm(A - A_low_rank)

print("\nApproximation Error:")
print(error)

print("\nIs approximation close to A?")
print(np.allclose(A, A_low_rank, atol=1))

if __name__ == "__main__":
    example_create_arrays()
    example_matrix_multiplication()
    example_transpose()
    example_determinant()
    example_inverse()
    example_solve_linear_system()
    example_eigen()
    example_norms()
    example_rank()
    example_svd()
```
